# Perceptron 구현

## Perceptron 수식
>### y = $\sigma(\displaystyle\sum_{i=1}^{3} w_{i}x_{i} + b)$

## Load modules



```python
import numpy as np

print("NumPy Version :{}".format(np.__version__))
```

## Input, Weight, Bias



```python
# Input, Weight, Bias
X = np.array([1,2,3])
W = np.array([4,5,6])
B = 10
```

## Multiply and Summation

```python
# Element-wise multiply
mul = W * X
print("mul = W * X : {}".format(mul))

# Summation array and Add bias
s = np.sum(mul) + B
print("s = {}".format(s))
```

## Activation Function

```python
# Activation Function : Step Function
y = np.float32(s > 10)
print("y = {}".format(y))
```
